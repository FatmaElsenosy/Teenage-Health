```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sb
from sklearn.impute import SimpleImputer
%matplotlib inline
```

```python
df=pd.read_csv("Teen_Mental_Health_Dataset.csv")
df
```

```python
df.describe()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 6))
sns.scatterplot(x='daily_social_media_hours', y='anxiety_level', data=df)
plt.title('Daily Social Media Hours vs. Anxiety Level')
plt.xlabel('Daily Social Media Hours')
plt.ylabel('Anxiety Level')
plt.show()

correlation = df['daily_social_media_hours'].corr(df['anxiety_level'])
print(f"Correlation between Daily Social Media Hours and Anxiety Level: {correlation:.2f}")
```
